```sql
SELECT DISTINCT library_date
FROM ml.library_snapshots
ORDER BY library_date;

WITH ss0808 AS (
  SELECT ls1.persistent_id, ls1.playcount
  FROM ml.library_snapshots ls1
  WHERE library_date = '2024-08-08'
)
, diffs AS (
  SELECT  IFNULL(ll.playcount,0) - IFNULL(ss0808.playcount, 0) as countdiff
    , ss0808.playcount as oldcount
    , ll.playcount, ll.album, ll.artist, ll.track_name, ll.persistent_id
  FROM ss0808
    JOIN ml.library_latest ll ON ss0808.persistent_id = ll.persistent_id
)
SELECT  *
  --concat('{"', persistent_id , '", ', oldcount, '}')
from diffs
where countdiff < 0
```
